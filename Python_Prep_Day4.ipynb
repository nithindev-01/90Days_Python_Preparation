#1. Check whether a number is even or odd

```python
num = int(input("Enter the number: "))
if num%2==0:
  print("Even Number")
else:
  print("Odd Number")
```

#2. Calculate a power without using **

#3.Find the longest word in a sentence

```python
text = input("Enter text: ")
words = text.split()
print(words)
print(max(words, key = len))
```
