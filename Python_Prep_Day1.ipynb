#Q1. Check whether a number is even or odd without converting it to a string.


```python
n = int(input("Enter the number:"))
def check_num(n):
  if n%2==0:
    return "Even Number"
  else:
    return "Odd Number"

check_num(n) # Direct approach
```

#Q2. Check whether a number is even or odd without converting it to a string.- Optimised Approach


```python
#Optimised approach--> Using bitwise operator
# Bitwise AND operator along with and any number always extracts the last bit of the binary number,
# Any binary number ending with a 0 is a even number and ending with 1 is an odd number
n = int(input("Enter the number:"))
def check_bit_num(n):
  if n&1==0:
    return 'Even number'
  else:
    return 'Odd number'

check_bit_num(n)
```

#Q4. Check whether a string is a palindrome ignoring case — First solve it with a straightforward approach, then optimize it.

```python
text = str(input("Enter the text:"))
text = text.replace(' ','').lower()
def check_palindrome(text):
  rev = ''
  for i in text:
    rev = i+rev
  if text == rev:
    print("Is a palindrome")
  else:
    print("Not a palindrome")
check_palindrome(text)
```

```python
#Optimised approach
text = str(input("Enter the text:"))
text = text.replace(' ','').lower()
def check_palindrome(text):
  text1 = text[::-1]
  if text == text1:
    print("Is a palindrome")
  else:
    print("Not a palindrome")
check_palindrome(text)
```
